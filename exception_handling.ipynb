```python
num = 100

while True:
    deno_input = input("Enter any number (or type 'stop'): ")

    # 1. Check if user wants to stop
    if deno_input.lower() == "stop":
        print("OK, Stopping...")
        break

    # 2. Handle the math and potential errors
    try:
        deno = int(deno_input)
        ans = num / deno
        print(f"Result: {ans}")
        
    except ZeroDivisionError:
        print("Zero is not a valid input, please retry.")
        
    except ValueError:
        print("Using text/strings is not logical, please use numbers.")
```
