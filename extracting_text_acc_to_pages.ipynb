```python
"""
Extract text from pages 68 to 140 of a PDF file.
Saves the extracted text to "68_to_140_pages.txt".
"""
 
import sys
from pathlib import Path


# ── Configuration ──────────────────────────────────────────────────────────────
PDF_FILE_PATH = r"C:\Users\User\Downloads\Demand Side Sales 101.pdf"   # <-- change this to your path
START_PAGE    = 165
END_PAGE      = 183
OUTPUT_FILE   = "ch7_demand_side_sales_101.txt"
# ──────────────────────────────────────────────────────────────────────────────
 
 
def extract_text_pypdf(pdf_path: str, start: int, end: int) -> str:
    """Extract text using pypdf (0-indexed pages internally)."""
    from pypdf import PdfReader
 
    reader = PdfReader(pdf_path)
    total  = len(reader.pages)
    print(f"  Total pages in PDF: {total}")
 
    actual_end = min(end, total)
    if start > total:
        raise ValueError(f"Start page {start} exceeds total pages ({total}).")
 
    texts = []
    for page_num in range(start - 1, actual_end):   # convert to 0-based index
        page = reader.pages[page_num]
        text = page.extract_text() or ""
        texts.append(f"--- Page {page_num + 1} ---\n{text}")
 
    return "\n\n".join(texts)
 
 
def extract_text_pdfplumber(pdf_path: str, start: int, end: int) -> str:
    """Fallback: extract text using pdfplumber (better layout awareness)."""
    import pdfplumber
 
    texts = []
    with pdfplumber.open(pdf_path) as pdf:
        total = len(pdf.pages)
        print(f"  Total pages in PDF: {total}")
 
        actual_end = min(end, total)
        if start > total:
            raise ValueError(f"Start page {start} exceeds total pages ({total}).")
 
        for page_num in range(start - 1, actual_end):  # 0-based
            page = pdf.pages[page_num]
            text = page.extract_text() or ""
            texts.append(f"--- Page {page_num + 1} ---\n{text}")
 
    return "\n\n".join(texts)
 
 
def main():
    pdf_path = PDF_FILE_PATH
 
    # Basic validation
    if not Path(pdf_path).exists():
        print(f"Error: file not found → {pdf_path}")
        sys.exit(1)
    if not pdf_path.lower().endswith(".pdf"):
        print(f"Warning: '{pdf_path}' does not have a .pdf extension.")
 
    print(f"PDF      : {pdf_path}")
    print(f"Pages    : {START_PAGE} – {END_PAGE}")
    print(f"Output   : {OUTPUT_FILE}")
    print()
 
    # Try pypdf first; fall back to pdfplumber if needed
    text = ""
    try:
        print("Extracting with pypdf …")
        text = extract_text_pypdf(pdf_path, START_PAGE, END_PAGE)
    except ImportError:
        print("  pypdf not available, trying pdfplumber …")
        try:
            text = extract_text_pdfplumber(pdf_path, START_PAGE, END_PAGE)
        except ImportError:
            print("Neither pypdf nor pdfplumber is installed.")
            print("Install one:  pip install pypdf   OR   pip install pdfplumber")
            sys.exit(1)
    except Exception as e:
        print(f"  pypdf failed ({e}), trying pdfplumber …")
        text = extract_text_pdfplumber(pdf_path, START_PAGE, END_PAGE)
 
    if not text.strip():
        print("\nWarning: no text was extracted. The PDF may be scanned/image-based.")
        print("Consider using OCR (e.g. pytesseract) for scanned documents.")
 
    # Save output
    with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
        f.write(text)
 
    char_count = len(text)
    word_count = len(text.split())
    print(f"\nDone! Extracted {char_count:,} characters / {word_count:,} words.")
    print(f"Saved to: {OUTPUT_FILE}")
 
 
if __name__ == "__main__":
    main()
```
