```python
import webbrowser
from pathlib import Path
from urllib.parse import urlparse
```

```python
FILE_NAME = r"C:\Users\User\OneDrive\Desktop\links.txt"
```

```python
def make_url(line: str) -> str | None:
    line = line.strip()

    # Ignore empty lines
    if not line:
        return None

    # Add https:// if the user omitted the protocol
    if not line.startswith(("http://", "https://")):
        line = "https://" + line

    # Basic URL validation
    parsed = urlparse(line)
    if not parsed.netloc:
        return None

    return line
```

```python
def main():
    file_path = Path(FILE_NAME)

    if not file_path.exists():
        print(f"File not found: {FILE_NAME}")
        return

    urls = []

    with file_path.open("r", encoding="utf-8") as file:
        for line in file:
            url = make_url(line)
            if url:
                urls.append(url)

    if not urls:
        print("No valid URLs found.")
        return

    print(f"Opening {len(urls)} links...")

    for url in urls:
        print(f"Opening: {url}")
        webbrowser.open_new_tab(url)
```

```python
if __name__ == "__main__":
    main()
```

```python
import subprocess
from pathlib import Path
import os


FILE_NAME = "links.txt"

# Common Chrome installation locations
CHROME_PATHS = [
    r"C:\Program Files\Google\Chrome\Application\chrome.exe",
    r"C:\Program Files (x86)\Google\Chrome\Application\chrome.exe",
    rf"C:\Users\{__import__('os').getuser()}\AppData\Local\Google\Chrome\Application\chrome.exe",
]


def find_chrome():
    for path in CHROME_PATHS:
        if Path(path).exists():
            return path
    return None


def main():
    chrome = find_chrome()

    if not chrome:
        print("Google Chrome was not found.")
        return

    file_path = Path(FILE_NAME)

    if not file_path.exists():
        print(f"{FILE_NAME} was not found.")
        return

    with file_path.open("r", encoding="utf-8") as file:
        links = [line.strip() for line in file if line.strip()]

    for link in links:
        if not link.startswith(("http://", "https://")):
            link = "https://" + link

        subprocess.Popen([chrome, "--new-tab", link])

    print(f"Opened {len(links)} links in Google Chrome.")


if __name__ == "__main__":
    main()
    
```
