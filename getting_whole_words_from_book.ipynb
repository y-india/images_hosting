```python
from PyPDF2 import PdfReader
import re

```

```python
import os
os.getcwd()
```

```python
# Input PDF file
pdf_path = "unlimited_memory.pdf"
```

```python
# Output text file
output_file = "whole_book_words.txt"
```

```python
# Read PDF
reader = PdfReader(pdf_path)

all_text = []
```

```python
# Loop through all pages
for page in reader.pages:
    text = page.extract_text()
    if text:
        all_text.append(text)
```

```python
# Join all pages into one string
final_text = "\n".join(all_text)

# Write to file
with open(output_file, "w", encoding="utf-8") as f:
    f.write(final_text)

print("Done. Text saved to whole_book_words.txt")
```

```python
with open("whole_book_words.txt", "r", encoding="utf-8") as f:
    text = f.read()
```

```python
pattern = r"(C\s*H\s*A\s*P\s*T\s*E\s*R\s+\d+\s*\.\s*(?:[A-Z]\s*)+)"
```

```python
parts = re.split(pattern, text)
```

```python
chapters = []

for i in range(1, len(parts), 3):
    title = parts[i]
    content = parts[i+1] if i+1 < len(parts) else ""
    chapters.append(title + content)
```

```python
group_size = 5

for i in range(0, len(chapters), group_size):
    group = chapters[i:i+group_size]
    
    file_text = "\n\n".join(group)
    
    file_name = f"chapters_{i//group_size + 1}.txt"
    
    with open(file_name, "w", encoding="utf-8") as f:
        f.write(file_text)

print("Done.")
```

```python
# chapters list already created from previous step

parts = {
    "part_1.txt": chapters[0:5],    # Chapter 1–5
    "part_2.txt": chapters[5:10],   # Chapter 6–10
    "part_3.txt": chapters[10:16]   # Chapter 11–16
}

for file_name, group in parts.items():
    file_text = "\n\n".join(group)
    
    with open(file_name, "w", encoding="utf-8") as f:
        f.write(file_text)

print("Done. 3 parts created.")
```
